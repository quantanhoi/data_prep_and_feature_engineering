### Question 1.2: Transform to Tidy Format (20 points)

**Instructions:** Write code to convert this data to tidy format. The final dataset should have these columns:
- product
- category
- year
- quarter
- sales

```python
# YOUR CODE HERE
# Transform sales_wide to sales_tidy

sales_tidy = pd.melt(
    sales_wide,
    id_vars = ['product', 'category'],
    var_name = 'year_quarter',
    value_name = 'sales'
    
)
sales_tidy[['year', 'quarter']] = sales_tidy['year_quarter'].str.split('_Q', expand=True) # _Q means split at _Q, drop _Q
sales_tidy = sales_tidy.drop(columns=['year_quarter'], axis=1)
sales_tidy['year'] = sales_tidy['year'].astype(int)
sales_tidy['quarter'] = sales_tidy['quarter'].astype(int)

# Display your result
print("Tidy Sales Data:")
print(sales_tidy)
print(f"\nShape: {sales_tidy.shape}")
```

**Instructions:** Convert this nested JSON structure to a tidy DataFrame where:
- Each cast member is a separate row
- Movie information is included with each cast member
- Final columns should be: movie_id, title, genre, year, director, actor

**HINT:** The 'cast' field contains a **list of strings** (not objects). You'll still use `pd.json_normalize()` with `record_path`, but the result will create a column named '0' that you need to rename to 'actor'!

```python
# YOUR CODE HERE
# Transform movies_data to movies_tidy

movies_tidy = pd.json_normalize(
    movies_data['movies'],
    record_path = 'cast',
    meta = ['movie_id', 'title', 'year', 'genre', 'director']
)
movies_tidy = movies_tidy.rename(columns={movies_tidy.columns[0]: 'actor'})

# Display your result
print("Tidy Movie Cast Data:")
print(movies_tidy)
print(f"\nShape: {movies_tidy.shape if movies_tidy is not None else 'None'}")
```

### Key Takeaways

Remember the three principles of tidy data:
1. Each variable forms a column
2. Each observation forms a row
3. Each type of observational unit forms a table

**Nested Arrays in JSON:**
- **Objects** (Task 3): `[{"key": "value"}, ...]` → Multiple columns from each object
- **Simple values** (Task 5): `["value1", "value2", ...]` → Single column (rename from '0')

When you encounter untidy data in the real world, ask yourself:
- Are variables hidden in column names?
- Are multiple variables crammed into one column?
- Do I have nested structures that should be flattened?
- Am I mixing different types of entities in one table?
